# Бақылаусыз оқыту


**Бақылаусыз оқыту** (`Unsupervised Learning`) — белгіленбеген деректермен жұмыс істейтін машиналық оқыту түрі. Мұнда деректердің дайын жауаптары немесе класс атаулары берілмейді.

**Модель адамның көмегінсіз:**

- бір-біріне ұқсас деректерді топтастырады;
- деректер арасындағы жасырын заңдылықтарды анықтайды;
- деректердің ішкі құрылымын зерттейді.

**Бақылаусыз оқыту:**

- кластерлеу (`clustering`);
- өлшемділікті азайту (`dimensionality reduction`);
- ассоциациялық ережелерді анықтау (`association rule learning`);
- деректерді топтау және қысу;
- қалыптан тыс деректерді анықтау (`anomaly detection`)

сияқты тапсырмаларда қолданылады.


{418216F6-3CF5-4B71-894B-7990A026030D}.png

Суретте бастапқы деректер ретінде пілдер, түйелер және сиырлар берілген. Бұл деректерде жануарлардың атаулары немесе кластары алдын ала көрсетілмеген.

- Түсіндіру кезеңінде модель деректердің дайын белгілері жоқ екенін анықтайды және олардың ұқсастықтарын іздейді.
- Алгоритм деректердегі жасырын заңдылықтар мен ұқсастықтарды табады.
- Өңдеу кезеңінде алгоритм жануарларды ортақ ерекшеліктеріне қарай топтастырады.
- Нәтижесінде пілдер, түйелер және сиырлар жеке кластерлерге бөлінеді.

## Бақылаусыз оқыту қалай жұмыс істейді?

### 1. Белгіленбеген деректерді жинау
Алдын ала берілген класс атаулары жоқ деректер жиналады.

Мысалы: жануарлардың атауы көрсетілмеген әртүрлі суреттер.

### 2. Алгоритмді таңдау

Мақсатқа байланысты тиісті алгоритм таңдалады:

- кластерлеу үшін — K-Means;
- ассоциациялық ережелерді анықтау үшін — Apriori;
- өлшемділікті азайту үшін — PCA.

### 3. Модельді бастапқы деректермен оқыту

Белгіленбеген деректер алгоритмге беріледі. Алгоритм олардың арасындағы ұқсастықтарды, байланыстарды және жасырын құрылымдарды іздейді.

### 4. Деректерді топтау немесе түрлендіру

Алгоритм адамның көмегінсіз ұқсас деректерді кластерлерге біріктіреді немесе олардың өлшемін азайтады.

Мысалы: пілдерді, түйелерді және сиырларды бөлек топтарға орналастырады.

### 5. Нәтижені талдау және пайдалану

Алынған кластерлер мен заңдылықтар талданады. Оларды:
- деректерді визуализациялау;
- аномалияларды анықтау;
- басқа модельдерге кіріс деректері ретінде беру
үшін пайдалануға болады.

## Бақылаусыз оқыту алгоритмдері

Алгоритмдердің негізгі үш түрі бар.

### 1. Кластерлеу алгоритмдері

**Кластерлеу** (`Clustering`) — белгіленбеген деректерді ұқсастықтарына қарай бірнеше топқа, яғни кластерге бөлу әдісі.

Мұнда модельге дайын класс атаулары берілмейді. Алгоритм деректердің ерекшеліктерін салыстырып, олардың арасындағы жасырын ұқсастықтар мен байланыстарды өзі анықтайды.

**Кластерлеу:**

- белгілері ұқсас деректерді бір топқа біріктіреді;
- бастапқы деректердегі табиғи топтарды анықтайды;
- тұтынушыларды сегменттеуде, аномалияларды анықтауда және деректерді ұйымдастыруда қолданылады;
- дайын жауаптарсыз, тек кіріс деректерімен жұмыс істейді;
- деректердің құрылымын түсінуге және шешім қабылдауға көмектеседі.

**Кең таралған кластерлеу алгоритмдері:**

- **K-means** — нүктелерді бір-біріне жақындығына қарай алдын ала берілген \(K\) кластерге бөледі.
- **Иерархиялық кластерлеу** — топтарды біртіндеп біріктіру немесе бөлу арқылы ағаш тәрізді құрылым құрады.
- **DBSCAN** — нүктелер тығыз орналасқан аймақтардан кластерлер құрады, ал жеке орналасқан нүктелерді шу немесе аномалия деп есептейді.
- **Mean Shift** — деректер нүктелерін ең тығыз аймақтарға қарай жылжыту арқылы кластерлерді анықтайды.
- **Спектрлік кластерлеу** — нүктелер арасындағы байланыстарды граф көмегімен талдап, деректерді топтарға бөледі.

### 2. Ассоциациялық ережелерді анықтау

**Ассоциациялық ережелерді анықтау** (`Association Rule Learning`) — үлкен деректер жиынындағы элементтер арасындағы жасырын байланыстарды табатын бақылаусыз оқыту әдісі.

Алгоритм байланыстарды **«егер..., онда...»** түріндегі ережелермен көрсетеді.

**Мысалы:**

> Егер сатып алушы нан алса, онда сүтті де сатып алуы мүмкін.

**Бұл әдіс:**

- бірге жиі кездесетін элементтерді анықтайды;
- олардың арасындағы байланысты көрсететін ережелерді табады;
- сатып алушылардың қандай өнімдерді бірге алатынын зерттейді;
- акциялар ұйымдастыруға және қосымша өнім ұсынуға көмектеседі.

**Кең таралған алгоритмдер:**

- **Apriori** — бірге жиі кездесетін элементтер жиынын кезең-кезеңімен іздейді.
- **FP-Growth** — кандидаттық жиындарды құрмай-ақ, жиі кездесетін комбинацияларды жылдам анықтайды. Ол `Apriori` алгоритміне қарағанда тиімдірек болуы мүмкін.
- **Eclat** — элементтер жиындарының қиылысуын пайдаланып, жиі кездесетін комбинацияларды табады.
- **Ағашқа негізделген алгоритмдер** — деректерді ағаш құрылымына орналастырып, үлкен деректер жиынын тиімді өңдейді.

### 3. Өлшемділікті азайту

**Өлшемділікті азайту** (`Dimensionality Reduction`) — бастапқы ақпаратты мүмкіндігінше сақтай отырып, деректердегі белгілер санын қысқарту әдісі.

Мысалы, деректер жиынында 100 белгі болса, олардың ішіндегі ең маңызды ақпаратты сақтап, белгілер санын 10-ға дейін азайтуға болады.

**Бұл әдіс:**

- көптеген белгілерді аз әрі маңызды белгілерге айналдырады;
- деректердегі негізгі заңдылықтарға назар аударуға көмектеседі;
- күрделі деректерді талдау мен визуализациялауды жеңілдетеді;
- есептеу уақытын және жад шығынын азайтады;
- модель жұмысын жылдамдатып, артық оқыту қаупін төмендетеді.

**Кең таралған алгоритмдер:**

- **PCA** — өзара байланысы жоқ негізгі компоненттерді құрып, деректердің өлшемін азайтады.
- **NMF** — деректерді теріс емес қарапайым бөліктерге жіктейді.
- **LLE** — жақын орналасқан нүктелер арасындағы байланысты сақтай отырып, өлшемділікті азайтады.
- **Isomap** — деректердің жалпы құрылымын және нүктелер арасындағы күрделі қашықтықтарды сақтайды.

## Қолданылу салалары

- **Клиенттерді сегменттеу** — сатып алу әрекеті немесе демографиялық белгілері ұқсас клиенттерді топтастырады.
- **Аномалияларды анықтау** — алаяқтықты, кибершабуылды немесе жабдық ақауын білдіретін қалыптан тыс деректерді табады.
- **Ұсыным жүйелері** — пайдаланушының әрекеті мен қызығушылығына қарай тауар, фильм немесе музыка ұсынады.
- **Кескіндер мен мәтіндерді топтау** — ұқсас суреттерді немесе құжаттарды бір топқа біріктіреді.
- **Әлеуметтік желілерді талдау** — пайдаланушылар арасындағы қауымдастықтар мен үрдістерді анықтайды.

## Артықшылықтары

- Белгіленбеген бастапқы деректермен жұмыс істейді.
- Деректерді қолмен белгілеуге кететін уақыт пен еңбекті үнемдейді.
- Адам бірден байқай алмайтын жасырын заңдылықтарды анықтайды.
- Үлкен және көпөлшемді деректерді өңдей алады.
- Алдын ала мысалдарсыз аномалияларды табуға көмектеседі.

## Қиындықтары

- Шу мен шектен тыс мәндер (`outliers`) модель нәтижесін бұрмалауы мүмкін.
- Модель маңызды заңдылықтардың орнына кездейсоқ шуды үйреніп, артық оқытуы мүмкін.
- Дайын жауаптар болмағандықтан, алгоритмді қажетті нәтижеге бағыттау қиын.
- Алынған кластерлерді түсіндіру қиын болуы мүмкін.
- Табылған топтар нақты өмірдегі категорияларға сәйкес келе бермейді.

# K-Means кластерлеу әдісіне кіріспе


**K-Means** — белгіленбеген деректерді ұқсастықтарына қарай бірнеше кластерге бөлетін мұғалімсіз оқыту алгоритмі.

**Бұл әдіс:**

- белгіленбеген деректердегі табиғи топтарды анықтайды;
- нүктелерді кластер орталықтарына дейінгі қашықтық бойынша топтастырады;
- клиенттерді сегменттеуде, кескіндерді қысуда және жасырын заңдылықтарды анықтауда қолданылады;
- реттелмеген бастапқы деректердің құрылымын түсінуге көмектеседі.

## K-Means қалай жұмыс істейді?

Әрбір нысан бірнеше белгімен сипатталған деректер жиыны берілсін.

Мысалы, клиентті:

- жасы;
- табысы;
- сатып алу сомасы

сияқты белгілермен көрсетуге болады.

**K-Means алгоритмінің міндеті** — осы нысандарды ұқсастықтарына қарай топтарға бөлу.

Алгоритм атауындағы:

- **K** — құрылатын кластерлердің саны;
- **Means** — әр кластердің орташа мәні, яғни орталығы.

Мысалы, `K = 3` болса, алгоритм деректерді **үш кластерге** бөледі.

## K-Means кластерлеуінің қолданылуы

K-Means алгоритмі қарапайым, жылдам және тиімді болғандықтан, көптеген салаларда қолданылады.

1. **Деректерді сегменттеу**

   Деректерді ерекшеліктеріне қарай жеке топтарға бөледі. Мысалы, компания клиенттерді сатып алу әрекеті немесе веб-сайтты пайдалану ерекшелігі бойынша топтастыра алады.

2. **Кескіндерді қысу**

   Түстері ұқсас пиксельдерді бір кластерге біріктіру арқылы кескіндегі түстер санын азайтады. Соның нәтижесінде кескінді сақтау және өңдеу жеңілдейді.

3. **Аномалияларды анықтау**

   Кластерлерден алыс орналасқан немесе ешбір топқа сәйкес келмейтін нүктелерді анықтайды. Мұндай нүктелер қалыптан тыс деректер немесе аномалиялар болуы мүмкін.

4. **Құжаттарды кластерлеу**

   Табиғи тілді өңдеуде мазмұны ұқсас құжаттар мен мақалаларды бір топқа біріктіреді. Бұл жаңалықтарды тақырыптарға бөлу және ұсыным жүйелерін құру кезінде қолданылады.

5. **Үлкен деректер жиынын ұйымдастыру**

   Үлкен көлемдегі деректерді ұқсастықтарына қарай шағын топтарға бөледі. Бұл деректерді талдау мен басқаруды жеңілдетеді.

## K-Means кластерлеуін жүзеге асыру

Бұл мысалда `make_blobs` көмегімен жасанды деректер жиынын құрып, K-Means алгоритмінің деректерді кластерлерге қалай бөлетінін көрсетеміз.

### 1-қадам. Қажетті кітапханаларды импорттау

Келесі кітапханалар қолданылады:

- **NumPy** — сандармен жұмыс істеу және қашықтықтарды есептеу үшін;
- **Matplotlib** — деректер мен кластерлеу нәтижесін графикте көрсету үшін;
- **Scikit-learn** — `make_blobs` функциясының көмегімен жасанды деректер жиынын құру үшін.



```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
```

### 2-қадам. Жасанды деректер жиынын құру

`make_blobs` функциясының көмегімен жасанды деректер жиынын құрамыз.

- `n_samples=500` — 500 дерек нүктесін жасайды;
- `n_features=2` — әр нүктеде екі белгі болады;
- `centers=3` — деректер үш топтың айналасында құрылады;
- `X[:, 0]` — барлық нүктенің бірінші белгісін алады;
- `X[:, 1]` — барлық нүктенің екінші белгісін алады;
- `plt.scatter()` — нүктелерді екі өлшемді графикте көрсетеді;
- `plt.show()` — графикті экранға шығарады.


Мұндағы `y` нүктелердің бастапқы топтарын сақтайды, бірақ K-Means оқыту кезінде оны пайдаланбайды. Себебі K-Means — бақылаусыз оқыту алгоритмі.

```python
X,y = make_blobs(n_samples = 500,n_features = 2,centers = 3,random_state = 23)

fig = plt.figure(0)
plt.grid(True)
plt.scatter(X[:,0],X[:,1])
plt.show()
```

### 3-қадам. `StandardScaler` көмегімен белгілерді масштабтау

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X = scaler.fit_transform(X)
```

### 4-қадам. Кездейсоқ центроидтарды бастапқы орнату

K-Means кластерлеуі үшін центроидтарды кездейсоқ таңдаймыз.

- `np.random.seed(23)` — кодты әр іске қосқанда бірдей кездейсоқ нәтижелердің алынуын қамтамасыз етеді.
- `for` циклі екі өлшемді деректер үшін мәндері `−2` мен `2` аралығындағы `k` кездейсоқ центроидты құрады.

```python
k = 3

clusters = {}
np.random.seed(23)

for idx in range(k):
    center = 2*(2*np.random.random((X.shape[1],))-1)
    points = []
    cluster = {
        'center' : center,
        'points' : []
    }

    clusters[idx] = cluster

clusters
```

### 5-қадам. Деректер нүктелері мен бастапқы центроидтарды көрсету

Енді деректер нүктелері мен кездейсоқ таңдалған бастапқы центроидтарды графикке шығарамыз.

- `plt.grid()` — графикке тор сызықтарын қосады.
- `plt.scatter(center[0], center[1], marker='*', c='red')` — кластер орталығын қызыл жұлдызша (`*`) түрінде көрсетеді.

```python
plt.scatter(X[:,0],X[:,1])
plt.grid(True)
for i in clusters:
    center = clusters[i]['center']
    plt.scatter(center[0],center[1],marker = '*',c = 'red')
plt.show()
```

### 6-қадам. Евклид қашықтығын анықтау

Әрбір дерек нүктесін ең жақын центроидқа тағайындау үшін қашықтықты есептейтін функцияны анықтаймыз.

- `np.sqrt()` — санның немесе массив элементтерінің квадрат түбірін есептейді.
- `np.sum()` — массивтегі барлық элементтердің немесе көрсетілген ось бойындағы элементтердің қосындысын есептейді.

{92FFFA2D-EFDA-4438-BAC2-10DD4D7E4963}.png

```python
def distance(p1,p2):
    return np.sqrt(np.sum((p1-p2)**2))
```

### 7-қадам. Нүктелерді тағайындау және центроидтарды жаңарту функцияларын құру

Келесі кезеңде әрбір нүктені ең жақын центроидқа тағайындайтын және әр кластердегі нүктелердің орташа мәні бойынша центроидтарды жаңартатын функциялар құрамыз.

- `dist.append(dis)` — есептелген қашықтықты `dist` тізіміне қосады.
- `curr_cluster = np.argmin(dist)` — ең кіші қашықтықтың индексін тауып, нүктеге ең жақын кластерді анықтайды.
- `new_center = points.mean(axis=0)` — кластердегі барлық нүктенің орташа мәнін есептеп, жаңа центроидты анықтайды.

```python
def assign_clusters(X, clusters):
    for idx in range(X.shape[0]):
        dist = []

        curr_x = X[idx]

        for i in range(k):
            dis = distance(curr_x,clusters[i]['center'])
            dist.append(dis)
        curr_cluster = np.argmin(dist)
        clusters[curr_cluster]['points'].append(curr_x)
    return clusters

def update_clusters(X, clusters):
    for i in range(k):
        points = np.array(clusters[i]['points'])
        if points.shape[0] > 0:
            new_center = points.mean(axis =0)
            clusters[i]['center'] = new_center

            clusters[i]['points'] = []
    return clusters
```

### 8-қадам. Дерек нүктелерінің кластерін болжау

Соңғы центроидтар негізінде әрбір дерек нүктесінің қай кластерге жататынын анықтайтын функция құрамыз.

- `pred.append(np.argmin(dist))` — ең жақын кластердің индексін, яғни қашықтығы ең аз центроидтың нөмірін `pred` тізіміне қосады.

```python
def pred_cluster(X, clusters):
    pred = []
    for i in range(X.shape[0]):
        dist = []
        for j in range(k):
            dist.append(distance(X[i],clusters[j]['center']))
        pred.append(np.argmin(dist))
    return pred
```

### 9-қадам. Кластер орталықтарын тағайындау, жаңарту және болжау

Нүктелерді кластерлерге тағайындау және центроидтарды жаңарту әрекеттері кластер орталықтары тұрақталғанша немесе итерациялардың ең үлкен санына жеткенше қайталанады.

- `old_centers` — центроидтардың жаңартуға дейінгі мәндерін сақтайды.
- `assign_clusters(X, clusters)` — әрбір дерек нүктесін ең жақын центроидқа тағайындайды.
- `update_clusters(X, clusters)` — кластердегі нүктелердің орташа мәні бойынша әр центроидты қайта есептейді.
- `np.allclose(old_centers, new_centers)` — ескі және жаңа центроидтардың мәндері шамамен бірдей болғанын тексереді. Егер бірдей болса, алгоритм жұмысын ерте тоқтатады.
- `pred_cluster(X, clusters)` — тұрақталған центроидтар бойынша әр дерек нүктесінің соңғы кластерін анықтайды.

```python
# Repeat until convergence or maximum iterations
max_iters = 100

for _ in range(max_iters):
    old_centers = [clusters[i]['center'].copy() for i in range(k)]

    clusters = assign_clusters(X, clusters)
    clusters = update_clusters(X, clusters)

    new_centers = [clusters[i]['center'] for i in range(k)]

    if np.allclose(old_centers, new_centers):
        break

pred = pred_cluster(X, clusters)
```

### 10-қадам. Дерек нүктелері мен кластер орталықтарын көрсету

Соңында дерек нүктелерін болжанған кластерлеріне сәйкес әртүрлі түспен, ал жаңартылған центроидтарды бөлек белгімен графикке шығарамыз.

- `center = clusters[i]['center']` — ағымдағы кластердің орталығын, яғни центроидын алады.
- `plt.scatter(center[0], center[1], marker='^', c='red')` — кластер орталығын графикте қызыл үшбұрыш (`^`) түрінде көрсетеді.

```python
plt.scatter(X[:,0],X[:,1],c = pred)
for i in clusters:
    center = clusters[i]['center']
    plt.scatter(center[0],center[1],marker = '^',c = 'red')
plt.show()
```

## K-Means алгоритмінің қиындықтары

- **Кластерлер санын (`K`) дұрыс таңдау:** деректерді қанша кластерге бөлу керектігін алдын ала анықтау қиын болуы мүмкін.
- **Бастапқы центроидтарға тәуелділік:** соңғы нәтиже центроидтардың бастапқы кездейсоқ орналасуына байланысты өзгеруі мүмкін.
- **Күрделі пішінді кластерлер:** K-Means кластерлерді шамамен дөңгелек және өлшемдері бірдей деп қабылдайды. Сондықтан пішіні немесе тығыздығы әртүрлі топтарды дұрыс анықтамауы мүмкін.
- **Шектен тыс мәндерге сезімталдық:** аномальды нүктелер центроидты өзіне қарай ығыстырып, кластерлеу нәтижесін бұрмалауы мүмкін.

# Негізгі компоненттер әдісі (PCA)

**PCA** (`Principal Component Analysis`) — деректердегі ең маңызды ақпаратты мүмкіндігінше сақтай отырып, белгілер санын азайтатын өлшемділікті қысқарту әдісі.

PCA өзара байланысты көптеген белгіні саны аз, бір-бірімен байланыспайтын жаңа белгілерге — **негізгі компоненттерге** айналдырады. Нәтижесінде күрделі деректерді талдау, визуализациялау және модельде қолдану жеңілдейді.


{B1707D1C-8A01-4C07-9A20-05ADCBC40BEC}.png

## Негізгі компоненттер әдісі қалай жұмыс істейді?

PCA сызықтық алгебра көмегімен бастапқы белгілерді **негізгі компоненттер** деп аталатын жаңа белгілерге түрлендіреді.

Ол ковариациялық матрица негізінде:

- **меншікті векторларды** — деректердің негізгі бағыттарын;
- **меншікті мәндерді** — әр бағыттың маңыздылығын

есептейді. Содан кейін меншікті мәндері ең үлкен компоненттерді таңдап, деректерді соларға проекциялайды. Нәтижесінде белгілер саны азайып, деректер қарапайым түрде көрсетіледі.

> **Ескерту:** PCA деректер ең көп өзгеретін бағыттарға басымдық береді. Себебі өзгеріс неғұрлым көп болса, сол бағытта маңызды ақпарат соғұрлым көп болады.

Деректерді аспандағы ретсіз орналасқан жұлдыздар сияқты елестетейік. PCA осы нүктелерді негізгі заңдылықтарын жоғалтпай көруге мүмкіндік беретін ең тиімді бағыттарды анықтайды.

### 1-қадам. Деректерді стандарттау

Белгілер әртүрлі өлшем бірліктері мен масштабтарға ие болуы мүмкін. Мысалы, адамның жасы ондаған сандармен, ал жалақысы жүздеген мың сандармен беріледі.

Сондықтан PCA қолданар алдында әр белгі стандартталады:

- орташа мәні — `0`;
- стандартты ауытқуы — `1`.

Стандарттау формуласы:

{A8E55E22-1617-4F57-8DFB-2C3C342D3ECC}.png

**Мұндағы:**

- X — бастапқы мән;
- mu — белгінің орташа мәні;
- sigma — белгінің стандартты ауытқуы;
- Z— стандартталған мән.

## PCA әдісін жүзеге асыру

PCA деректердегі өзгергіштіктің (`variance`) ең үлкен бөлігін сақтай отырып, белгілер санын азайту үшін сызықтық түрлендіруді қолданады. Әдіс келесі қадамдардан тұрады.

### 1-қадам. Қажетті кітапханаларды импорттау

Нәтижелерді есептеу және визуализациялау үшін келесі кітапханаларды импорттаймыз:

- `Pandas` — кестелік деректермен жұмыс істеу үшін;
- `NumPy` — сандық есептеулер үшін;
- `Scikit-learn` — PCA әдісін және деректерді стандарттауды қолдану үшін;
- `Seaborn` және `Matplotlib` — нәтижелерді график түрінде көрсету үшін.

```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns
```

### 2-қадам. Үлгілік деректер жиынын құру

Үш сандық белгісі бар шағын деректер жиынын құрамыз:

- бойы (`Height`);
- салмағы (`Weight`);
- жасы (`Age`).

Сонымен қатар адамның жынысын көрсететін `Gender` категориялық белгісі беріледі. `Gender` PCA есептеуіне тікелей қатыспайды, бірақ алынған нәтижені графикте әртүрлі түспен көрсету үшін қолданылуы мүмкін.

```python
data = {
    'Height': [170, 165, 180, 175, 160, 172, 168, 177, 162, 158],
    'Weight': [65, 59, 75, 68, 55, 70, 62, 74, 58, 54],
    'Age': [30, 25, 35, 28, 22, 32, 27, 33, 24, 21],
    'Gender': [1, 0, 1, 1, 0, 1, 0, 1, 0, 0]  # 1 = Male, 0 = Female
}
df = pd.DataFrame(data)
print(df)
```

### 3-қадам. Деректерді стандарттау

Бой мен жас сияқты белгілердің өлшемдері мен масштабтары әртүрлі болғандықтан, деректерді стандарттаймыз.

Стандарттаудан кейін әр белгінің:
- орташа мәні — 0;
- стандартты ауытқуы — 1 болады.
Бұл үлкен мәндері немесе өлшем бірлігіне байланысты бір белгінің PCA нәтижесіне шамадан тыс әсер етуіне жол бермейді.

```python
X = df.drop('Gender', axis=1)
y = df['Gender']

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
# Деректерді бөлу
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.3, random_state=42)

# PCA
pca = PCA(n_components=2)
X_train_pca = pca.fit_transform(X_train)
X_test_pca = pca.transform(X_test)

# Лог регрессияны оқыту
model = LogisticRegression()
model.fit(X_train_pca, y_train)

# Болжау
y_pred = model.predict(X_test_pca)
print("Предсказанные значения:", y_pred)
```

```python
cm = confusion_matrix(y_test, y_pred)

plt.figure(figsize=(5,4))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=['Female', 'Male'], yticklabels=['Female', 'Male'])
plt.xlabel('Predicted Label')
plt.ylabel('True Label')
plt.title('Confusion Matrix')
plt.show()
```

```python
# Prepare data for visualization
scaler_viz = StandardScaler()
X_scaled_viz = scaler_viz.fit_transform(X)

pca_viz = PCA(n_components=2)
X_pca_viz = pca_viz.fit_transform(X_scaled_viz)

y_numeric = pd.factorize(y)[0]

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.scatter(
    X_scaled_viz[:, 0], X_scaled_viz[:, 1],
    c=y_numeric, cmap='coolwarm', edgecolor='k', s=80
)
plt.xlabel('Original Feature 1')
plt.ylabel('Original Feature 2')
plt.title('Before PCA: Using First 2 Standardized Features')
plt.colorbar(label='Target classes')

plt.subplot(1, 2, 2)
plt.scatter(
    X_pca_viz[:, 0], X_pca_viz[:, 1],
    c=y_numeric, cmap='coolwarm', edgecolor='k', s=80
)
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.title('After PCA: Projected onto 2 Principal Components')
plt.colorbar(label='Target classes')

plt.tight_layout()
plt.show()
```
