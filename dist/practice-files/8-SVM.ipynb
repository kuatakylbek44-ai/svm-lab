# SVM алгоритмі

**SVM (Support Vector Machine)** көбінесе жіктеу есептерінде қолданылады.

**Мысалы:**

- хат — спам немесе спам емес;
- ісік — қатерлі немесе қатерсіз;
- кескін — мысық немесе ит.

### SVM қалай жұмыс істейді?

Екі классқа жататын нүктелер бар деп елестетейік. Оларды бірнеше түзу сызықпен бөлуге болады. Бірақ SVM кез келген сызықты емес, екі класстан да мүмкіндігінше алыс орналасқан **ең тиімді шекараны** таңдайды.

Бұл шекара:

- екі өлшемді кеңістікте — **түзу сызық**;
- үш өлшемді кеңістікте — **жазықтық**;
- жоғары өлшемді кеңістікте — **гипержазықтық** деп аталады.

Сызықтық шешім шекарасының теңдеуі:

\[
w^Tx+b=0
\]

Мұнда \(w\) шекараның бағытын, ал \(b\) оның орналасуын анықтайды.

### Тірек векторлар деген не?

**Тірек векторлар** — шешім шекарасына ең жақын орналасқан нүктелер.

Дәл осы нүктелер шекараның қай жерде өтетінін анықтайды. Сондықтан басқа нүктелерге қарағанда тірек векторлардың SVM үшін маңызы жоғары.

### Margin деген не?

**Margin — екі класс арасындағы қауіпсіз аралық.**

SVM осы аралықты барынша үлкейтуге тырысады. Аралық үлкен болса, модель жаңа деректерді дұрыс жіктеуге жақсырақ бейімделеді.

Қарапайым мысал: SVM екі топтың ортасынан барынша кең жол салуға тырысады. Жолдың ортасы — шешім шекарасы, ал жолдың шеттеріне жақын нүктелер — тірек векторлар.

### Hard Margin және Soft Margin

**Hard Margin** деректерді қатесіз бөлуге тырысады. Бірде-бір нүкте шекараны бұзбауы керек. Ол тек деректер таза және толық ажыратылатын кезде қолайлы.

**Soft Margin** кейбір нүктелердің қате жіктелуіне немесе аралықтың ішіне түсуіне мүмкіндік береді. Шынайы деректерде шу мен шектен тыс мәндер кездесетіндіктен, көбіне Soft Margin қолданылады.

### `C` параметрі нені басқарады?

`C` — қате жіктеуге қаншалықты қатаң қарау керектігін анықтайтын параметр.

- **`C` үлкен болса** — модель қателерді қатты жазалайды, оқыту деректерін дәлірек бөлуге тырысады. Бірақ шамадан тыс үйренуі мүмкін.
- **`C` кіші болса** — кейбір қателерге мүмкіндік береді және кеңірек аралықты таңдайды. Модель жаңа деректерге жақсырақ жалпылануы мүмкін.

Яғни, `C` аралықтың кеңдігі мен жіктеу қателерінің арасында тепе-теңдік орнатады.

### Деректер түзу сызықпен бөлінбесе не болады?

Кейбір деректерді бір түзу сызықпен бөлу мүмкін емес. Мұндай жағдайда SVM **ядролық функцияны — kernel** қолданады.

Kernel деректерді жоғары өлшемді кеңістікке ауыстырғандай әсер береді. Сол кеңістікте оларды түзу шекарамен бөлу оңайырақ болады.

Негізгі ядролар:

- `linear` — сызықтық деректер үшін;
- `poly` — полиномдық шекара үшін;
- `rbf` — күрделі және сызықтық емес деректер үшін.

Бұл әдіс **kernel trick** деп аталады.

## SVM-нің негізгі идеясы

**Қысқаша айтқанда, SVM:**

1. кластарды бөлетін шекараны іздейді;
2. шекараға ең жақын тірек векторларды анықтайды;
3. тірек векторлар мен шекара арасындағы аралықты барынша үлкейтеді;
4. қажет болса, kernel арқылы сызықтық емес деректерді бөледі.

**Ең қарапайым түсінік:** SVM екі кластың арасынан ең кең және қауіпсіз бөлуші жолды табады.

### **Сызықтық SVM**

- Сызықтық SVM әртүрлі кластарға жататын деректер нүктелерін бөлу үшін сызықтық шешім шекарасын қолданады.
- Егер деректерді сызық арқылы нақты бөлуге болатын болса, сызықтық SVM өте тиімді жұмыс істейді.
- Бұл екі өлшемді кеңістікте деректерді бір түзу сызықпен, ал жоғары өлшемді кеңістікте гипержазықтықпен тиісті кластарға толық бөлуге болатынын білдіреді.
- Кластар арасындағы аралықты барынша үлкейтетін гипержазықтық шешім шекарасы болып табылады.

### **Сызықтық емес SVM**

- **Сызықтық емес SVM** деректерді түзу сызық арқылы екі класқа бөлу мүмкін болмаған жағдайда қолданылады.
- Ол сызықтық түрде бөлінбейтін деректерді өңдеу үшін ядролық функцияларды қолданады.
- Ядролық функциялар бастапқы деректерді жоғары өлшемді белгілер кеңістігіне түрлендіреді. Бұл кеңістікте деректер нүктелерін сызықтық түрде бөлуге болады.
- Түрлендірілген кеңістікте сызықтық SVM қолданылады. Соның нәтижесінде бастапқы кеңістіктегі сызықтық емес шешім шекарасы анықталады.

{C3AD25FE-BDD3-42B3-B67F-9E25FD48D7D3}.png

````python
from sklearn.datasets import load_breast_cancer
import matplotlib.pyplot as plt
from sklearn.inspection import DecisionBoundaryDisplay    # Модельдің екі класты қалай бөлгенін, яғни шешім шекарасын көрсету үшін қолданылады.
from sklearn.svm import SVC # Жіктеуге арналған Support Vector Classifier моделін импорттайды.

cancer = load_breast_cancer()
X = cancer.data[:, :2]    # Барлық жолдардан тек алғашқы екі белгіні аламыз
y = cancer.target   # Мақсатты класс мәндерін y айнымалысына сақтайды: 0 — қатерлі ісік, 1 — қатерсіз ісік

svm = SVC(kernel="linear", C=1)    # 'kernel' - кластарды түзу сызықпен бөлуге тырысады; `C=1` — аралықтың кеңдігі мен жіктеу қателерінің арасындағы тепе-теңдікті реттейді.
svm.fit(X, y)

DecisionBoundaryDisplay.from_estimator(
        svm,
        X,
        response_method="predict",
        alpha=0.8,
        cmap="Pastel1",
        xlabel=cancer.feature_names[0],
        ylabel=cancer.feature_names[1],
    )

plt.scatter(X[:, 0], X[:, 1],
            c=y,
            s=20, edgecolors="k")
plt.show()
````

Бұл код **сүт безі қатерлі ісігі деректерін** пайдаланып, сызықтық SVM моделін үйретеді және екі класс арасындағы шешім шекарасын графикте көрсетеді.


```python
svm = SVC(kernel="linear", C=1)
```

Сызықтық SVM моделін құрады.

- `kernel="linear"` — кластарды түзу сызықпен бөлуге тырысады;
- `C=1` — аралықтың кеңдігі мен жіктеу қателерінің арасындағы тепе-теңдікті реттейді.

```python
svm.fit(X, y)
```

Модельді `X` белгілері мен `y` кластары бойынша оқытады. SVM екі класты ажырататын ең тиімді шешім шекарасын табады.

---

Оқытылған модельдің шешім аймақтарын графикке салады.

- `X` — графикте қолданылатын белгілер;

- `alpha=0.8` — фон түстерінің мөлдірлігін реттейді;
- `cmap="Pastel1"` — фон үшін жұмсақ түстер палитрасын қолданады;

- `ylabel` — Y осіне екінші белгінің атауын жазады.

## 2-мысал: Дөңгелек шешім шекарасы бар сызықтық емес SVM

Төменде дөңгелек шешім шекарасы бар сызықтық емес SVM

### **1. Кітапханаларды импорттау**

Алдымен деректерді құру, модельді оқыту, бағалау және нәтижелерді визуализациялау үшін қажетті кітапханаларды импорттаймыз.

````python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_circles
from sklearn.svm import SVC
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
````

### **2. Деректер жиынын құру және бөлу**

Бір орталықтан таралған шеңберлер түріндегі жасанды деректер жиынын құрамыз және оны оқыту мен тестілеу жиындарына бөлеміз.

````python
X, y = make_circles(n_samples=500, factor=0.5, noise=0.05, random_state=42)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
````

### **3. Сызықтық емес SVM моделін құру және оқыту**

Сызықтық емес заңдылықтарды өңдеу үшін **RBF ядросын** қолданатын SVM жіктеу моделін құрамыз және оны деректер жиынында оқытамыз.

````python
svm = SVC(kernel='rbf', C=1, gamma=0.5)  #RBF ядросы дөңгелек шешім шекараларын анықтауға мүмкіндік береді
svm.fit(X_train, y_train)
````

### **4. Болжам жасау және модельді бағалау**

Тестілік деректер үшін класс белгілерін болжаймыз және модельдің дәлдігін есептейміз.

````python
y_pred = svm.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)
print(f"Accuracy: {accuracy:.2f}")
````

### **5. Шешім шекарасын визуализациялау**

Деректер жиынында оқытылған сызықтық емес SVM моделінің шешім шекарасын көрсету үшін арнайы функция анықтаймыз.

````python
def plot_decision_boundary(X, y, model):
    x_min, x_max = X[:, 0].min() - 1, X[:, 0].max() + 1
    y_min, y_max = X[:, 1].min() - 1, X[:, 1].max() + 1
    xx, yy = np.meshgrid(np.arange(x_min, x_max, 0.01),
                         np.arange(y_min, y_max, 0.01))

    Z = model.predict(np.c_[xx.ravel(), yy.ravel()])
    Z = Z.reshape(xx.shape)

    plt.contourf(xx, yy, Z, alpha=0.8, cmap=plt.cm.Paired)
    plt.scatter(X[:, 0], X[:, 1], c=y, edgecolor='k', cmap=plt.cm.Paired)
    plt.title("Сызықты емес SVM RBF Kernel")
    plt.show()

# Шешім шекарасын графикте көрсету
plot_decision_boundary(X, y, svm)
````
