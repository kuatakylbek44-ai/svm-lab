Клиенттердің отбасылық жағдайы, табысы, сатып алған тауарларының саны, тауар түрлері және басқа да сипаттамалары туралы ақпаратты қамтитын деректер жиынын жүктейміз.

Деректер жиыны дискте тұр

```python
df = pd.read_csv('new.csv')
df.head()
```

### 3-қадам. Деректерді алдын ала өңдеу (Data Preprocessing)

Деректер жиыны туралы ақпарат алу, атап айтқанда, бос мәндерді (`null`), жазбалар санын және басқа да сипаттамаларды тексеру үшін `.info()` әдісін қолданамыз.

```python
df.info()
```

```python
df.describe().T
```

```python
for col in df.columns:
    temp = df[col].isnull().sum()
    if temp > 0:
        print(f'Column {col} contains {temp} null values.')
```

```python
df = df.dropna()
print("Total values in the dataset after removing the null values:", len(df))
```

```python
df.nunique()
```

Мұнда деректер жиынындағы кейбір бағандардың барлық жолдарында бірдей мәндер бар екенін байқауға болады. Сондықтан мұндай бағандар модельді құру үшін маңызды ақпарат бермейді.

Сонымен қатар, деректер жиынында клиенттің тіркелген күнін көрсететін `Dt_Customer` бағаны бар. Бұл бағандағы күнді үш бөлек бағанға бөлуге болады: күн (day), ай (month) және жыл (year).

```python
parts = df["Dt_Customer"].str.split("-", n=3, expand=True)
df["day"] = parts[0].astype('int')
df["month"] = parts[1].astype('int')
df["year"] = parts[2].astype('int')
```

Енді біз барлық маңызды белгілерді анықтадық. Сондықтан модельді құруға қажет емес `Z_CostContact`, `Z_Revenue` және `Dt_Customer` бағандарын деректер жиынынан алып тастай аламыз.

```python
df.drop(['Z_CostContact', 'Z_Revenue', 'Dt_Customer'],
        axis=1,
        inplace=True)
```

### 4-қадам. Деректерді визуализациялау және талдау (Data Visualization and Analysis)

Деректерді визуализациялау (Data Visualization) — ақпарат пен деректерді графиктер немесе диаграммалар түрінде көрнекі көрсету әдісі.

Бұл қадамда деректерді жақсырақ визуализациялау және талдау үшін бағандық диаграмманы (Bar Plot) және сандық жиілік диаграммасын (Count Plot) қолданамыз.

```python
floats, objects = [], []
for col in df.columns:
    if df[col].dtype == object:
        objects.append(col)
    elif df[col].dtype == float:
        floats.append(col)

print(objects)
print(floats)
```

`object` деректер түріне жататын бағандар үшін сандық жиілік диаграммасын (Count Plot) құру мақсатында төмендегі кодты қолданамыз.

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Үш диаграмманы орналастыру
plt.figure(figsize=(12, 7))

plt.subplot(2, 2, 1)
sns.countplot(data=df, x='Education')
plt.xticks(rotation=0)

plt.subplot(2, 2, 2)
sns.countplot(data=df, x='Marital_Status')
plt.xticks(rotation=0)

plt.subplot(2, 2, 3)
sns.countplot(data=df, x='AcceptedCmp1')
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```

Деректер жиынындағы Marital_Status (отбасылық жағдайы) бағанының әрбір категориясы қанша рет кездесетінін value_counts() әдісі арқылы тексерейік.

```python
df['Marital_Status'].value_counts()
```

Енді белгілерді `Response` бағанының мәндерімен салыстырып, олардың арасындағы байланысты қарастырайық.

```python
plt.subplots(figsize=(15, 10))
for i, col in enumerate(objects):
    plt.subplot(2, 2, i + 1)

    df_melted = df.melt(id_vars=[col], value_vars=['Response'], var_name='hue')
    sb.countplot(x=col, hue='value', data=df_melted)
plt.show()
```

Белгілерді сандық кодтау (Label Encoding) — категориялық мәндерді модель түсініп, өңдей алатындай сандық мәндерге түрлендіру әдісі.

```python
for col in df.columns:
    if df[col].dtype == object:
        le = LabelEncoder()
        df[col] = le.fit_transform(df[col])
```

Жылу картасы (Heatmap) — деректер жиынындағы әртүрлі белгілер арасындағы корреляцияны (өзара байланысты) визуализациялаудың тиімді әдістерінің бірі.

Корреляция шегін 0,8 деп белгілейік. Бұл белгілер арасындағы күшті байланысты анықтауға мүмкіндік береді.

```python
plt.figure(figsize=(15, 15))
sb.heatmap(df.corr() > 0.8, annot=True, cbar=False)
plt.show()
```

Стандарттау (Standardization) — белгілердің масштабын біріздендіру әдісі және деректерді дайындаудың (Feature Engineering) маңызды бөлігі.

Бұл әдіс деректерді машиналық оқыту моделіне ыңғайлы масштабқа келтіріп, модельдің оларды тиімді меңгеруіне көмектеседі.

Стандарттау нәтижесінде:

- Орташа мән (Mean) — 0-ге тең болады.
- Стандартты ауытқу (Standard Deviation) — 1-ге тең болады.

```python
scaler = StandardScaler()
data = scaler.fit_transform(df)
```

### 5-қадам. Сегменттеу (Segmentation)

Бұл қадамда t-таралымды стохастикалық көршілерді ендіру (t-distributed Stochastic Neighbor Embedding — t-SNE) әдісін қолданамыз.

t-SNE — көпөлшемді деректерді визуализациялауға көмектесетін өлшемділікті азайту әдісі.

Бұл әдіс дерек нүктелері арасындағы ұқсастықтарды бірлескен ықтималдықтарға түрлендіреді және оларды төмен өлшемді кеңістікте мүмкіндігінше сақтауға тырысады. Осылайша, көпөлшемді деректерді екі немесе үш өлшемді кеңістікте көрнекі түрде көрсетуге мүмкіндік береді.

```python
from sklearn.manifold import TSNE
model = TSNE(n_components=2, random_state=0)
tsne_data = model.fit_transform(df)
plt.figure(figsize=(7, 7))
plt.scatter(tsne_data[:, 0], tsne_data[:, 1])
plt.show()
```

Берілген деректерді екі өлшемді (2D) кеңістікте визуализациялау нәтижесінде бірнеше кластердің анық байқалатынын көруге болады.

Жазықтықтағы дерек нүктелерін әртүрлі топтарға бөлу үшін K-Means кластерлеу (K-Means Clustering) алгоритмін де қолдануға болады.

```python
error = []
for n_clusters in range(1, 21):
    model = KMeans(init='k-means++',
                   n_clusters=n_clusters,
                   max_iter=500,
                   random_state=22)
    model.fit(df)
    error.append(model.inertia_)
```

Мұндағы инерция (Inertia) — кластер ішіндегі әрбір дерек нүктесінен өз кластерінің орталығына (центроидқа) дейінгі қашықтықтардың квадраттарының қосындысы.

```python
plt.figure(figsize=(10, 5))
sb.lineplot(x=range(1, 21), y=error)
sb.scatterplot(x=range(1, 21), y=error)
plt.show()
```

Мұнда Elbow (шынтақ) әдісін қолдану арқылы кластерлердің оңтайлы саны K = 6 екенін анықтауға болады. Себебі K = 6 мәнінен кейін инерция көрсеткіші айтарлықтай төмендемейді.

```python
model = KMeans(init='k-means++',
               n_clusters=5,
               max_iter=500,
               random_state=22)
segments = model.fit_predict(df)
```

Нүктелік диаграмма (Scatterplot) K-Means кластерлеу алгоритмі арқылы құрылған барлық 6 кластерді визуализациялау үшін қолданылады.

```python
plt.figure(figsize=(7, 7))

df_tsne = pd.DataFrame({'x': tsne_data[:, 0], 'y': tsne_data[:, 1], 'segment': segments})

sb.scatterplot(x='x', y='y', hue='segment', data=df_tsne)
plt.show()
```

Мұнда клиенттерді 5 кластерге бөлгенімізді көруге болады. Осы кластерлер негізінде сатып алу мінез-құлқы ұқсас клиенттерге бағытталған маркетингтік стратегияларды тиімдірек қолдана аламыз.

Сонымен қатар, әрбір клиенттер тобына жекелендірілген жарнамалар ұсынып, бизнестің дамуы мен өсуіне ықпал ететін негізделген шешімдер қабылдауға болады.
